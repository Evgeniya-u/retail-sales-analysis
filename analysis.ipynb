## Импорт библиотек

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-darkgrid')
%matplotlib inline
```

## Загрузка данных

```python
df = pd.read_csv('sales_data.csv')
df.head()
```

## Первичный анализ

```python
df.info()
df.describe()
```

```python
df.isnull().sum()
```

## Визуализация

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Sales'], bins=20, edgecolor='black')
plt.title('Распределение продаж')
plt.xlabel('Сумма продаж ($)')
plt.ylabel('Количество заказов')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df.groupby('Category')['Sales'].sum().sort_values().plot(kind='barh')
plt.title('Продажи по категориям')
plt.xlabel('Общие продажи ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[['Sales', 'Quantity', 'Profit']].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Корреляция')
plt.show()
```

## Анализ по категориям

```python
df.groupby('Category')['Sales'].mean()
```

```python
df.groupby('Category')['Profit'].sum()
```

1. **Technology** — наиболее прибыльная категория (>50% выручки)
